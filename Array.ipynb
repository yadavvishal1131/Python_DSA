# Array 

# 1.Find the largest element in an array.

```python
# Find the largest element in an array.
arr = [10,25,8,45,32]

largest = arr[0]

for num in arr:
    if num > largest:
        largest = num
print("largest element",largest)
```

```python
## Using indexing.
arr = [10,25,8,45,32]

largest = arr[0]

for i in range(len(arr)):
    if arr[i] > largest:
        largest = arr[i]
print(largest)
```

# 2.Find the second largest element.

```python
arr = [10,25,8,45,32]

largest = arr[0]
second_largest = arr[0]

for num in arr:
    if num > largest:
       second_largest = largest
       largest = num
    elif num > second_largest and num != largest:
         second_largest = num

# print(largest)
print(second_largest)

```

```python
# if the value is negative in the array then we can use.

arr = [10,25,8,45,32]

largest = float('-inf')
second_largest = float('-inf')

for num in arr:
    if num > largest:
       second_largest = largest
       largest = num
    elif num > second_largest and num != largest:
         second_largest = num

# print(largest)
print(second_largest) 
```

# 3.Find the smallest element.

```python
arr = [10, 25, 8, 45, 32]

smallest = arr[0]

for num in arr:
    if num < smallest:
        smallest = num

print(smallest)
```

```python
#  By using indexing
arr = [10, 25, 8, 45, 32]

smallest = arr[0]

for i in range(len(arr)):
    if arr[i] < smallest:
        smallest = arr[i]

print(smallest)
```

 # 4.Reverse an array.

```python
# using function 
arr = [10, 25, 8, 45, 32]

arr.reverse()
print(arr)
```

```python
arr = [10, 25, 8, 45, 32]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print(arr)

```
